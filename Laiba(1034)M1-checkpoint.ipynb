# Loan Approval Prediction System

#  Problem Statement

In the financial sector, particularly in banking, loan approvals are a critical yet complex process. Banks and financial institutions need to assess loan applications efficiently and accurately to ensure minimal risk of default while maintaining customer satisfaction. Traditionally, this process involves manual reviews of applicants' profiles, which can be time-consuming, prone to human error, and inconsistent.

```python
import gradio as gr
```

#  In this we give the gradio model4 that is kNeighbour Classifier for the predictions

it takes the independent attributes as the input from the user and generate the reesults accordingly

```python
# Define the prediction function
def predict_loan_status(gender, married, dependents, education, self_employed, CoapplicantIncomelog,ApplicantIncomelog,
                        Total_Incomelog, loan_amountlog, 
                        loan_amount_termLog, credit_history, property_area):
    # Preprocess input features into the required format
    input_features = [
        float(gender), float(married), float(dependents), float(education), float(self_employed),
        float(CoapplicantIncomelog),float(ApplicantIncomelog),float(Total_Incomelog), float(loan_amountlog),
        float(loan_amount_termLog), float(credit_history), float(property_area)
    ]
    input_features = np.array(input_features).reshape(1, -1)  # Reshape for model
    prediction = model4.predict(input_features)
    return "Loan Approved" if prediction[0] == 1 else "Loan Rejected"

# Define Gradio interface
demo = gr.Interface(
    fn=predict_loan_status,
    inputs=[
        gr.Number(label="Gender (1: Male, 0: Female)"),
        gr.Number(label="Married (1: Yes, 0: No)"),
        gr.Number(label="Dependents"),
        gr.Number(label="Education (1: Graduate, 0: Not Graduate)"),
        gr.Number(label="Self Employed (1: Yes, 0: No)"),
        gr.Number(label="CoapplicantIncomelog"),
        gr.Number(label="ApplicantIncomelog"),
        gr.Number(label="Total_IncomeLog"),
        gr.Number(label="Loan AmountLog"),
        gr.Number(label="Loan Amount TermLog"),
        gr.Number(label="Credit History (1: Good, 0: Bad)"),
        gr.Number(label="Property Area (1: Urban, 0: Rural)"),
        
    ],
    outputs=gr.Textbox(label="Loan Status"),
    title="Loan Eligibility Prediction",
    description="Provide all required details to predict loan approval status."
)

# Launch the app
demo.launch()
```
